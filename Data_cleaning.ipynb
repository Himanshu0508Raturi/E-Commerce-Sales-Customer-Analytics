# E-Commerce Data Preprocessing & Quick Analysis

This notebook performs simple **data cleaning and basic analysis** on:

- Orders.csv
- Customers.csv
- Products.csv
- Categories.csv

The cleaned data can then be used for SQL EDA.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
orders = pd.read_csv('./data/Orders.csv')
customers = pd.read_csv('./data/Customers.csv')
products = pd.read_csv('./data/Products.csv')
categories = pd.read_csv('./data/Categories.csv')

print('Orders:', orders.shape)
print('Customers:', customers.shape)
print('Products:', products.shape)
print('Categories:', categories.shape)
```

```python
print('Orders')
display(orders.head())

print('Customers')
display(customers.head())

print('Products')
display(products.head())

print('Categories')
display(categories.head())
```

```python
# Check missing value
print(f"Orders Missing values: \n{orders.isnull().sum()}")

print(f"\nCustomers Missing values: \n{customers.isnull().sum()}")

print(f"\nProducts Missing values: \n{products.isnull().sum()}")

print(f"\nCategories Missing values: \n{categories.isnull().sum()}")
```

```python
# Filling Null Values.
orders['Aging'] = orders['Aging'].fillna(orders['Aging'].median())
orders['Sales'] = orders['Sales'].fillna(orders['Sales'].median())
orders['Quantity'] = orders['Quantity'].fillna(orders['Quantity'].median())
orders['Discount'] = orders['Discount'].fillna(orders['Discount'].median())
orders['Shipping_Cost'] = orders['Shipping_Cost'].fillna(orders['Shipping_Cost'].median())
orders['Order_Priority'] = orders['Order_Priority'].fillna(
    orders['Order_Priority'].mode()[0]
)
```

Parent_Category_Id should be as NULL/NaN, because that preserves the original hierarchy information.

```python
# Check duplicate rows
print('Orders duplicates:', orders.duplicated().sum())
print('Customers duplicates:', customers.duplicated().sum())
print('Products duplicates:', products.duplicated().sum())
print('Categories duplicates:', categories.duplicated().sum())
# If present remove them using .drop_duplicates() methods
```

```python
# Remove unnecessary spaces from important text fields.
orders['Gender'] = orders['Gender'].str.strip()
orders['Device_Type'] = orders['Device_Type'].str.strip()
orders['Customer_Login_type'] = orders['Customer_Login_type'].str.strip()
orders['Product'] = orders['Product'].str.strip()
orders['Product_Category'] = orders['Product_Category'].str.strip()
orders['Order_Priority'] = orders['Order_Priority'].str.strip()
orders['Payment_Method'] = orders['Payment_Method'].str.strip()

customers['Gender'] = customers['Gender'].str.strip()
customers['Device_Type'] = customers['Device_Type'].str.strip()
customers['Customer_Login_type'] = customers['Customer_Login_type'].str.strip()

products['Product_Name'] = products['Product_Name'].str.strip()
products['Product_Category'] = products['Product_Category'].str.strip()

categories['Category_Name'] = categories['Category_Name'].str.strip()

print('Text cleaning completed.')
```

```python
# Convert the order date into a proper date and important numeric columns into numeric types.

orders['Order_Date'] = pd.to_datetime(orders['Order_Date'] , dayfirst=True , errors='coerce')
orders['Sales'] = pd.to_numeric(orders['Sales'], errors='coerce')
orders['Quantity'] = pd.to_numeric(orders['Quantity'], errors='coerce')
orders['Discount'] = pd.to_numeric(orders['Discount'], errors='coerce')
orders['Profit'] = pd.to_numeric(orders['Profit'], errors='coerce')
orders['Shipping_Cost'] = pd.to_numeric(orders['Shipping_Cost'], errors='coerce')
orders['Aging'] = pd.to_numeric(orders['Aging'], errors='coerce')

print(orders[['Order_Date', 'Sales', 'Quantity', 'Discount', 'Profit', 'Shipping_Cost']].dtypes)
```

```python
# Checking for negative Values
print('Negative Quantity:', (orders['Quantity'] < 0).sum())
print('Negative Sales:', (orders['Sales'] < 0).sum())
print('Negative Discount:', (orders['Discount'] < 0).sum())
print('Negative Shipping Cost:', (orders['Shipping_Cost'] < 0).sum())
print('Negative Profit:', (orders['Profit'] < 0).sum())
```

```python
orders.to_csv('Orders.csv', index=False)
customers.to_csv('Customers.csv', index=False)
products.to_csv('Products.csv', index=False)
categories.to_csv('Categories.csv', index=False)

print('CSV files saved successfully.')
```

```python
from pathlib import Path
import sqlite3

db_path = Path("Ecommerce_SQL_DATAProject1.db")
conn = sqlite3.connect(db_path)

orders.to_sql("orders", conn, if_exists="replace", index=False)
customers.to_sql("customers", conn, if_exists="replace", index=False)
products.to_sql("products", conn, if_exists="replace", index=False)
categories.to_sql("categories", conn, if_exists="replace", index=False)



conn.close()
print("Created:", db_path)
```
